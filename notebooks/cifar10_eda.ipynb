```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from tensorflow.keras.datasets import cifar10

# 1. Load Dataset
(x_train, y_train), (x_test, y_test) = cifar10.load_data()

# CIFAR-10 label names
class_names = [
    "airplane", "automobile", "bird", "cat", "deer",
    "dog", "frog", "horse", "ship", "truck"
]

print("Train images:", x_train.shape)
print("Train labels:", y_train.shape)
print("Test images:", x_test.shape)
print("Test labels:", y_test.shape)
```

```python
plt.figure(figsize=(10, 4), dpi=200)
for i in range(10):
    plt.subplot(2, 5, i+1)
    plt.imshow(x_train[i])
    plt.title(class_names[y_train[i][0]])
    plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 4), dpi=200)
for i in range(10):
    plt.subplot(2, 5, i+1)
    plt.imshow(x_train[i], interpolation='nearest')
    plt.title(class_names[y_train[i][0]])
    plt.axis('off')
plt.tight_layout()
plt.show()
```

```python

unique, counts = np.unique(y_train, return_counts=True)
class_counts = dict(zip(unique.flatten(), counts))

plt.figure(figsize=(8, 4))
sns.barplot(x=[class_names[i] for i in class_counts.keys()], y=class_counts.values())
plt.title("Class Distribution in Training Set")
plt.xticks(rotation=45)
plt.show()
```

```python
# 4. Pixel statistics (mean and std per channel)

mean = np.mean(x_train, axis=(0,1,2)) / 255
std = np.std(x_train, axis=(0,1,2)) / 255

print("Mean per channel (R, G, B):", mean)
print("Std per channel (R, G, B):", std)
```

```python
# 5. Display pixel range of a sample image

sample_img = x_train[0]
print("Pixel min:", sample_img.min(), "Pixel max:", sample_img.max())

plt.imshow(sample_img)
plt.title(class_names[y_train[0][0]])
plt.axis('off')
plt.show()
```
